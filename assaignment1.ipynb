```python
import random
M1=[]
M2=[]
for i in range(5):
    row = []
    for j in range(10):
        row.append(random.randint(0,50))
    M1.append(row)
for i in range(10):
    row = []
    for j in range(5):
        row.append(random.randint(0,50))
    M2.append(row)
print(M1)
print(M2)
def Matrix_multip(M1,M2):
    result=[]
    for i in range(5):
        row=[]
        for j in range(5):
            s=0
            for k in range(10):
                s=s+M1[i][k]*M2[k][j]
            row.append(s)
        result.append(row)
    return result
result=Matrix_multip(M1,M2)
print(result)
```

```python
def Pascal_triangle(n):
    result=[]
    row=[1]
    result.append(row)
    row=[1,1]
    result.append(row)
    for i in range(2,n):
        row=[1]*(i+1)
        for j in range(1,i): 
             value=result[i-1][j-1]+result[i-1][j]
             row[j]=value
        result.append(row)
    return result
result=Pascal_triangle(100)
print(result)
```

```python
import random
moves=["double","add one RMB"]
action = random.choice(moves)
def money(n):
    Least_moves=[]
    for k in range(10000):
        a=1
        i=0
        while a<n:
            action = random.choice(moves)
            if action=="double":
                a=a*2
                i=i+1
            if action=="add one RMB":
                a=a+1
                i=i+1
        Least_moves.append(i)
    return Least_moves
x=money(1000)
print(min(x)) 
```

```python
def Find_expression(n) :
    digits="123456789"
    result = []
    def dfs(index, expression, total, current):
        if index == len(digits):
            if total+current==n:
                result.append(expression)
            return
        num=int(digits[index])
        dfs(index+1,expression+"+"+str(num),total+current,num)
        dfs(index+1,expression+"-"+str(num),total+current,-num)
        if current >= 0:
            new_current=current*10+num
        else:
            new_current=current*10-num
        dfs(index+1,expression+str(num),total,new_current)
    dfs(1,"1",0,1)
    return result
x=Find_expression(50)
print(x)
print("解的数量:", len(x))

for a in range(1,101):
    x=Find_expression(a)
    print("当x=",a,"解的数量:", len(x))


import matplotlib.pyplot as plt
x_values = []
y_values = []
for a in range(1, 101):
    x=Find_expression(a)
    x_values.append(a)
    y_values.append(len(x))
plt.plot(x_values, y_values)
plt.xlabel("x")
plt.ylabel("number of solutions")
plt.title("Number of solutions for x = 1 to 100")
plt.show()

plt.bar(x_values, y_values)
plt.xlabel("x")
plt.ylabel("number of solutions")
plt.title("Number of solutions for x = 1 to 100")
plt.show()
```
